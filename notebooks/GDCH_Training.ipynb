# GDCH End-to-End Training

This notebook preprocesses the OPO events and trains a GDCH model using the provided scripts.

```python
from gdch.data import process_raw_csv

process_raw_csv(
    input_path="dataframes/cleaned_data.csv",
    output_events_path="data/events.csv",
    output_metadata_path="data/metadata.json",
    output_opo_metadata_path="data/opo_metadata.csv",
    output_distance_path="data/distance.npy",
)
print("Processed events written to data/")
```

```python
from gdch.utils import load_config
from gdch.train import train_from_config

config = load_config("gdch/configs/base.json")
run_dir = train_from_config(config)
print(f"Run artifacts stored in: {run_dir}")
```
